Ejercicio 13

```python
def suma_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

numeros = [10, 20, 30, 40]
print(f"La suma de los elementos es: {suma_lista(numeros)}")
```

Ejercicio 14

```python
def maximo_lista(lista):
    mayor = lista[0]
    for elemento in lista:
        if elemento > mayor:
            mayor = elemento
    return mayor

numeros = [7, 3, 9, 2]
print(f"El elemento máximo es: {maximo_lista(numeros)}")
```

Ejercicio 15

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

suma_diagonal = 0
for i in range(len(matriz)):
    suma_diagonal += matriz[i][i]

print(f"La suma de la diagonal principal es: {suma_diagonal}")
```

Ejercicio 16

```python
def imprimir_promedio(lista):
    total = 0
    for elemento in lista:
        total += elemento
    promedio = total / len(lista)
    print(f"El promedio es: {promedio:.2f}")

calificaciones = [8, 9, 10, 7]
imprimir_promedio(calificaciones)
```

Ejercicio 18

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

numero = 5
print(f"El factorial de {numero} es: {factorial(numero)}")
```
