Calculadora de tiempo digital

```python
nombre = input("Ingresa tu nombre porfavor")
redes_sociales = float(input("Ingresa tus horas diarias en redes sociales :"))
mensajeria = float(input("Ingresa tus Horas diarias en mensajeria :"))
streaming = float(input("Ingresa tus Horas diarias en streaming :"))
Videojuegos= float(input("Ingresa tus Horas diarias en videojuegos :"))
estudio_en_linea= float(input("Ingresa tus Horas diarias en estudio en linea:"))
tiempo_total = ( redes_sociales + mensajeria + streaming + Videojuegos + estudio_en_linea )
porcentaje = (tiempo_total / 24) * 100
print (f"Usuario: {nombre}")
print (f"Tiempo acumulado : {tiempo_total}" , "horas")
print (f"Porcentaje del dia : {porcentaje:.2f}" , "%")
```
