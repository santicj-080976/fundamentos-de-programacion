Calculadora de tiempo digital

```python
nombre = input("pon tu nombre porfavor")
redes_sociales = float(input("pon tus horas diarias en las redes sociales porfa :"))
mensajeria = float(input("pon tus Horas diarias en la mensajeria porfa :"))
streaming = float(input("pon tus Horas diarias en el streaming porfa :"))
Videojuegos= float(input("pon tus Horas diarias en los videojuegos porfa :"))
estudio_en_linea= float(input("pon tus Horas diarias en el estudio en linea porfa :"))
tiempo_total = ( redes_sociales + mensajeria + streaming + Videojuegos + estudio_en_linea )
porcentaje = (tiempo_total / 24) * 100
print (f"tu usuario es : {nombre}")
print (f"El tiempo en el que estuviste en tu cel es  : {tiempo_total}" , "horas")
print (f"Tu porcentaje durante el día es: {porcentaje:.2f}" , "%")
```
