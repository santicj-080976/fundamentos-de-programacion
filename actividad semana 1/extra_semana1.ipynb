```python
total_cuenta = float(input(" pon el total de la cuenta :"))
porcentaje_propina = float(input("Pon la propina :"))
personas = int(input("cuantas personas? :"))

propina = porcentaje_propina / 100 * total_cuenta
total_con_propina = total_cuenta + propina
pagoxpersona = total_con_propina / personas
print(f"Propina : ${propina:.2f}")
print(f"total con propina : ${total_con_propina:.2f}")
print(f"Pago por persona : ${pagoxpersona:.2f}") 
```

```python
total_minutos = int(input("pon los  minutos: "))

dias = total_minutos // 1440
minutos_restantes = total_minutos % 1440
horas = minutos_restantes // 60
minutos = minutos_restantes % 60
print(f"{dias} días, {horas} horas, {minutos} minutos")
```
