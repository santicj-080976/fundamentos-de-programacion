Extra 1

```python
total = 0

while True:
    costo = float(input("Ingresa el costo del boleto: "))

    if costo == 0:
        print("Menor de 3 años, sin ningun cargo")
        continue

    total = total + costo

    if total >= 100:
        print("Aforo alcanzado, se detiene el registro")
        break

print(f"Total acumulado: ${total:.2f}")
```

Extra 2

```python
Visitantes = int(input("¿Cuantos visitantes van a ser?"))

adultos = 0
edad_acumulada = 0

for v in range(1, Visitantes+ 1):
    edad = int(input(f"Edad del visitante {v}: "))

    edad_acumulada = edad_acumulada + edad

    if edad >= 18:
        adultos = adultos + 1

Promedio_de_edad = edad_acumulada / Visitantes
print(f"Numero de visitantes : {Visitantes}")
print(f"Adultos: {adultos}")
print(f"Promedio de edad: {Promedio_de_edad:.0f}")
```
