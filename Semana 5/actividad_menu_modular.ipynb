```python
def Tuplas():
    numeros = (10, 20, 30, 40, 50)

    print(" Tuplas ")
    print("Tercer elemento:", numeros[2])

    try:
        numero1 = int(input("Pon un número adicional: "))
        numero2 = int(input("Ponotro número adicional: "))

        nueva_tupla = numeros + (numero1, numero2)
        print("Nueva tupla:", nueva_tupla)

        lista = list(nueva_tupla)
        lista.sort()

        print("Lista ordenada:", lista)
        print("total:", suma_tupla(nueva_tupla))

    except ValueError:
        print("Error: Debes ingresar números enteros.")
def suma_tupla(numeros):
    suma = 0
    for numero in numeros:
        suma += numero
    return suma

def diccionarios ():
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mía": "555-0103"
    }

    print(" Diccionarios ")

    nombre_nuevo = input("Pon el nombre del nuevo contacto: ")
    telefono_nuevo = input("Pon el teléfono del nuevo contacto: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("Nombres de los contactos:")

    for nombre in contactos.keys():
        print(nombre)

    nombre_buscar = input("Pon el nombre que quieres buscar: ")

    telefono = buscar_telefono(contactos, nombre_buscar)

    print("Teléfono:", telefono)

def buscar_telefono(contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")

def excepciones():
    print(" Excepciones ")

    try:
        numero1 = int(input("Pon el primer número entero: "))
        numero2 = int(input("Pon el segundo número entero: "))

        print("La suma total es:", numero1 + numero2)

        try:
            resultado = numero1 / numero2
            print("La división es:", resultado)

        except ZeroDivisionError:
            print("Error: No es posible dividir entre cero.")

    except ValueError:
        print("Error: Debes ingresar números enteros válidos.")





def string():
    mensaje = "Arriba  el cruz azul"

    print(" Strings ")

    print("Longitud del mensaje:", len(mensaje))
    print("En mayúsculas:", mensaje.upper())
    print("Texto reemplazado:", mensaje.replace("Arriba", "el America"))
    print("Palabras totales:", contar_palabras(mensaje))
def contar_palabras(texto):
    return len(texto.split())

while True:

    print(" Menu principal")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Finalizar")

    try:
        opcion = int(input("Selecciona opción: "))

        if opcion == 1:
            Tuplas()

        elif opcion == 2:
            diccionarios ()

        elif opcion == 3:
            excepciones()

        elif opcion == 4:
            string()

        elif opcion == 5:
            print("Programa finalizado.")
            break

        else:
            print("Opción no válida.")

    except ValueError:
        print("Error: Pon un número del menú.")
```
