Desafio extra 2

```python
candidatos = {
    "Rojo": 0,
    "Verde": 0,
    "Azul": 0
}

cantidad = int(input("¿Cuántos votos van a ser? "))

conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (Rojo, Verde o Azul): ")
    voto = voto.capitalize()
    if voto in candidatos:
        print(f"Voto para {voto}")
        candidatos[voto] += 1
        conteo += 1
    else:
        print("El voto no es válido")
ganador = ""
mayor = -1

for k, v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
    if v > mayor:
        mayor = v
        ganador = k
print(f"Ganador: {ganador} con {mayor} votos")
```

Desafio extra 5

```python
def mostrar_tupla():

    numeros = (8, 3, 12, 5, 1)

    lista = list(numeros)

    lista.sort()

    print("Tupla ordenada:", lista)


def buscar_telefono():

    contactos = {
        "Ana": "545-0890",
        "Luis": "565-0102",
        "Mía": "595-1023"
    }

    nombre = input("Nombre a buscar: ")

    print("Teléfono:", contactos.get(nombre, "Contacto no encontrado"))


def dividir():

    try:

        primero = int(input("Primer número: "))
        segundo = int(input("Segundo número: "))

        print("Resultado:", primero / segundo)

    except ValueError:

        print("Error: Ingresa números enteros válidos.")

    except ZeroDivisionError:

        print("Error: No se puede dividir entre cero.")


def analizar_mensaje():

    mensaje = input("Ingresa un mensaje: ")

    print("Longitud:", len(mensaje))

    print("Mayúsculas:", mensaje.upper())

    print("Reemplazo:", mensaje.replace("Python", "programación"))

    print("Palabras:", len(mensaje.split()))


while True:

    print(" Mini Menu ")
    print("1. Mostrar tupla ordenada")
    print("2. Buscar teléfono")
    print("3. Dividir números")
    print("4. Analizar mensaje")
    print("5. Finalizar")

    try:

        opcion = int(input("Selecciona una opción: "))

        if opcion == 1:

            mostrar_tupla()

        elif opcion == 2:

            buscar_telefono()

        elif opcion == 3:

            dividir()

        elif opcion == 4:

            analizar_mensaje()

        elif opcion == 5:

            print("Programa finalizado.")

            break

        else:

            print("Opción no válida.")

    except ValueError:

        print("Ingresa una opción numérica.")
```

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

nueva_calificacion1 = float(input("Nueva calificación 1: "))
nueva_calificacion2 = float(input("Nueva calificación 2: "))

nuevas_calificaciones = calificaciones + (nueva_calificacion1, nueva_calificacion2)

print("Tercera calificación:", nuevas_calificaciones[2])

lista = list(nuevas_calificaciones)

lista.sort(reverse=True)

def sumar_calificaciones(tupla):

    suma = 0

    for calificacion in tupla:

        suma += calificacion

    return suma

print("Nueva tupla:", nuevas_calificaciones)

print("Lista ordenada (mayor a menor):", lista)

print("total:", sumar_calificaciones(nuevas_calificaciones))
```

```python
agenda_contactos = {
    "Ana": "565-0101",
    "Luis": "565-0102",
    "Mía": "595-1027"
}

nombre = input("Nombre del nuevo contacto: ")

telefono = input("Numero del nuevo contacto: ")

agenda_contactos[nombre] = telefono

print("Contactos registrados:", ", ".join(agenda_contactos.keys()))

def buscar_telefono(agenda_contactos, nombre):

    return agenda_contactos.get(nombre, "Contacto no encontrado")

buscar = input("Nombre a buscar: ")

print(f"El teléfono de {buscar} es: {buscar_telefono(agenda_contactos, buscar)}")
```
